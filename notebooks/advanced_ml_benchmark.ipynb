# ADVANCED ML BENCHMARK: Gradient Boosting & Neural Network Attack Detection

## Objective
Comprehensive benchmark comparing 5 distinct Machine Learning and Deep Learning architectures for cyber attack detection:
1. **Logistic Regression** (Linear baseline)
2. **Random Forest Classifier** (Bagging ensemble baseline)
3. **XGBoost Classifier** (Extreme Gradient Boosting)

5. **Multi-Layer Perceptron (MLP)** (Deep Neural Network architecture)

- **Overall Metrics**: Accuracy, Weighted Precision, Recall, and F1-Score.
- **Minority Class Sensitivity**: Recall detection rates on rare attacks (`SQLI` and `PATH_TRAVERSAL`).
- **Inference Latency**: Execution latency per 10,000 predictions.
